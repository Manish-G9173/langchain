# HANDLING PDF CHALLENGES

🔴 Purpose of This Section PDFs are notoriously difficult to parse because they:
  - Store text in complex ways (not just simple text)
  - Can have formatting issues
  - May contain scanned images (requiring OCR)
  - Often have extraction artifacts

```python
# ============================================================================
# HANDLING PDF CHALLENGES
# ============================================================================
# PDFs are notoriously difficult to parse because they:
#   - Store text in complex ways (not just simple text)
#   - Can have formatting issues
#   - May contain scanned images (requiring OCR)
#   - Often have extraction artifacts
# ============================================================================

# ============================================================================
# EXAMPLE OF RAW PDF EXTRACTION
# ============================================================================
# This shows what raw text from a PDF looks like before cleaning
# Notice the excessive whitespace, newlines, and formatting artifacts
# ============================================================================

raw_pdf_text = """Company Financial Report

The financial performance for fiscal year 2023 shows significant growth in profitability.

Revenue increased by 25%.

The company's efficiency improved due to workflow optimization.

Page 1 of 10
"""

print("=" * 80)
print("RAW PDF EXTRACTION (BEFORE CLEANING):")
print("=" * 80)
print(repr(raw_pdf_text))
print("\n" + "-" * 80)

# ============================================================================
# PDF CLEANING FUNCTION
# ============================================================================
# This function addresses common PDF extraction issues:
# 1. Excessive whitespace (multiple spaces, newlines, tabs)
# 2. Ligature issues (common in PDF text extraction)
# 3. Formatting artifacts
# ============================================================================

def clean_text(text):
    """
    Clean raw PDF text by removing common extraction artifacts.
    
    Args:
        text (str): Raw text extracted from PDF
        
    Returns:
        str: Cleaned text with normalized whitespace and fixed ligatures
    """
    
    # Remove excessive whitespace
    # This converts multiple spaces, newlines, and tabs into single spaces
    # .split() splits on any whitespace and removes empty strings
    # " ".join() joins them back with single spaces
    text = " ".join(text.split())
    
    # Fix ligatures - common issue in PDF text extraction
    # Ligatures are characters that are combined into a single glyph
    # Common ligatures: fi, fl, ff, ffi, ffl, etc.
    # When extracted, they may appear as special characters or broken text
    text = text.replace("&", "fi")   # Fix 'fi' ligature
    text = text.replace("&", "fl")   # Fix 'fl' ligature
    # Note: The actual ligature characters may vary depending on the PDF
    
    return text

# ============================================================================
# APPLY THE CLEANING FUNCTION
# ============================================================================

cleaned = clean_text(raw_pdf_text)

print("=" * 80)
print("PDF TEXT CLEANING DEMONSTRATION:")
print("=" * 80)

print("\nBEFORE CLEANING:")
print("-" * 80)
print(repr(raw_pdf_text[:100]))  # Show first 100 characters
print("\nAFTER CLEANING:")
print("-" * 80)
print(repr(cleaned[:100]))       # Show first 100 characters

print("\n" + "=" * 80)
print("OUTPUT:")
print("=" * 80)

print("BEFORE:")
print(repr(raw_pdf_text[:100]))
print("\nAFTER:")
print(repr(cleaned[:100]))

# ============================================================================
# COMPARISON OF BEFORE AND AFTER
# ============================================================================

print("\n" + "=" * 80)
print("COMPARISON OF BEFORE AND AFTER CLEANING:")
print("=" * 80)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│  BEFORE CLEANING:                                                          │
│  ─────────────────────────────────────────────────────────────────────────  │
│  'Company Financial Report\\n\\n\\n The financial performance for fiscal   │
│  year 2024\\n shows significant gro'                                       │
│                                                                             │
│  Issues:                                                                    │
│  ❌ Multiple newlines (\\n\\n\\n)                                          │
│  ❌ Extra spaces (leading space)                                            │
│  ❌ Newline in middle of sentence (\\n shows)                              │
│  ❌ Inconsistent formatting                                                 │
│                                                                             │
│  AFTER CLEANING:                                                           │
│  ─────────────────────────────────────────────────────────────────────────  │
│  'Company Financial Report The financial performance for fiscal year       │
│  2024 shows significant growth in'                                         │
│                                                                             │
│  Improvements:                                                              │
│  ✅ Single spaces between all words                                        │
│  ✅ No leading/trailing whitespace                                         │
│  ✅ Text flows naturally without breaks                                    │
│  ✅ Consistent formatting                                                  │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ============================================================================
# COMMON PDF EXTRACTION ISSUES AND SOLUTIONS
# ============================================================================

print("=" * 80)
print("COMMON PDF EXTRACTION ISSUES AND SOLUTIONS:")
print("=" * 80)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│  ISSUE 1: EXCESSIVE WHITESPACE                                             │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: PDFs often have multiple spaces, newlines, and tabs              │
│  Solution: text = " ".join(text.split())                                   │
│  Example: "Hello    World\\n\\n\\nHow are you?" → "Hello World How are you?"│
│                                                                             │
│  ISSUE 2: LIGATURE ARTIFACTS                                                │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Common ligatures (fi, fl, ff) are often corrupted in extraction │
│  Solution: text = text.replace("&", "fi")                                  │
│  Example: "ﬁnancial" → "financial"                                         │
│                                                                             │
│  ISSUE 3: PAGE NUMBERS AND HEADERS                                         │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Page numbers, headers, footers get mixed with content           │
│  Solution: Use regex to remove patterns like "Page X of Y"                │
│  import re                                                                 │
│  text = re.sub(r'Page \\d+ of \\d+', '', text)                            │
│                                                                             │
│  ISSUE 4: HYPHENATION                                                       │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Words may be hyphenated across lines                             │
│  Solution: text = text.replace("-\\n", "")  # Join hyphenated words        │
│  Example: "perfor-\\nmance" → "performance"                                │
│                                                                             │
│  ISSUE 5: SPECIAL CHARACTERS                                                │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Em-dashes, en-dashes, curly quotes get converted to symbols     │
│  Solution: text = text.replace("–", "-")  # Convert to standard            │
│  text = text.replace("’", "'")   # Convert curly quotes                    │
│  text = text.replace("“", '"')   # Convert curly quotes                    │
│  text = text.replace("”", '"')   # Convert curly quotes                    │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ============================================================================
# ADVANCED CLEANING FUNCTION
# ============================================================================

print("=" * 80)
print("ADVANCED CLEANING FUNCTION:")
print("=" * 80)

print("""
Here's a more comprehensive cleaning function that handles multiple issues:

def advanced_clean_text(text):
    '''Comprehensive PDF text cleaning.'''
    import re
    
    # 1. Remove excessive whitespace
    text = " ".join(text.split())
    
    # 2. Fix common ligatures
    text = text.replace("&", "fi")
    text = text.replace("&", "fl")
    text = text.replace("&", "ff")
    
    # 3. Remove page numbers and headers
    text = re.sub(r'Page \\d+ of \\d+', '', text)
    text = re.sub(r'Page \\d+', '', text)
    
    # 4. Fix hyphenation (remove hyphens at line breaks)
    text = re.sub(r'(\\w+)-\\s+(\\w+)', r'\\1\\2', text)
    
    # 5. Fix common special characters
    text = text.replace("–", "-")
    text = text.replace("—", "-")
    text = text.replace("’", "'")
    text = text.replace("‘", "'")
    text = text.replace("“", '"')
    text = text.replace("”", '"')
    
    # 6. Remove multiple spaces again (just in case)
    text = " ".join(text.split())
    
    return text
""")

# ============================================================================
# SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("SUMMARY: PDF CLEANING BEST PRACTICES")
print("=" * 80)

print("""
✅ ALWAYS CLEAN PDF TEXTS BEFORE USE:
   → Raw PDF text is messy and inconsistent
   → Cleaning improves retrieval quality
   → Clean text is easier for LLMs to process

✅ START WITH SIMPLE CLEANING:
   → Remove excessive whitespace
   → Fix ligatures
   → Then handle specific issues

✅ KNOW YOUR PDF SOURCE:
   → Different PDF generators produce different artifacts
   → Test on a sample first
   → Adjust cleaning based on the output

✅ COMMON PDF CLEANING STEPS:
   1. Normalize whitespace
   2. Fix ligatures
   3. Remove page numbers
   4. Fix hyphenation
   5. Clean special characters
   6. Remove headers/footers

✅ FOR RAG APPLICATIONS:
   → Clean text before chunking
   → Clean text before embedding
   → Better input = Better output
""")

print("\n" + "=" * 80)
print("✅ PDF CLEANING DEMONSTRATION COMPLETE!")
print("=" * 80)
```
