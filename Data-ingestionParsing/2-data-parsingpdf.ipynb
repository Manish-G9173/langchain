```python
"""
EXTRACTOR 1: TEXT EXTRACTION ONLY

This module is responsible for extracting text content from PDF files and splitting them
into manageable chunks for further processing (like embedding or RAG applications).
"""

# Import necessary classes from LangChain
from langchain_core.documents import Document  # Standard document structure used across LangChain
from typing import List  # Type hint for lists
from langchain_text_splitters import (
    CharacterTextSplitter,          # Splits text by a specific character (e.g., "\n\n")
    RecursiveCharacterTextSplitter, # Splits hierarchically (paragraphs → sentences → words) - MOST COMMON
    TokenTextSplitter               # Splits by number of tokens (useful for LLM token limits)
)
from langchain_community.document_loaders import PyPDFLoader  # Specialized PDF loader

class TextExtractor:
    """
    A class to extract text from PDF files and split them into chunks.
    
    This extractor follows a two-step process:
    1. Load PDF pages using PyPDFLoader
    2. Split the extracted text into overlapping chunks using RecursiveCharacterTextSplitter
    
    Attributes:
        chunk_size (int): Maximum size of each text chunk in characters
        chunk_overlap (int): Number of characters to overlap between consecutive chunks
        text_splitter (RecursiveCharacterTextSplitter): The splitter instance with custom separators
    """
    
    def __init__(self, chunk_size=1000, chunk_overlap=100):
        """
        Initialize the TextExtractor with custom chunking parameters.
        
        Args:
            chunk_size (int): Target size for each text chunk (default: 1000 characters)
            chunk_overlap (int): Overlap between chunks to maintain context (default: 100 characters)
        
        The chunk_overlap is crucial for maintaining context across chunk boundaries,
        especially important for tasks like question answering where the answer might
        be split across two chunks.
        """
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        
        # Initialize a recursive text splitter
        # Recursive splitter tries to split at paragraph breaks first, then sentences, then words
        # This preserves semantic meaning better than simple character splitting
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,  # Maximum characters per chunk
            chunk_overlap=chunk_overlap,  # Characters to overlap between chunks
            separators=["\n\n", "\n", " ", ""]  # Priority order of separators to try
            # "\n\n" → paragraph breaks (highest priority - best semantic split)
            # "\n" → line breaks
            # " " → word boundaries
            # "" → character level (last resort - splits anywhere)
        )
    
    def extract_text(self, pdf_path: str) -> List[Document]:
        """
        Extract text from a PDF file and return it as a list of Document chunks.
        
        The extraction process:
        1. Load all pages from the PDF
        2. For each page:
           a. Clean the text (remove extra whitespace)
           b. Skip if page has less than 50 characters (likely empty or just images)
           c. Create chunks with page metadata
        3. Return all chunks as Document objects
        
        Args:
            pdf_path (str): File path to the PDF document
            
        Returns:
            List[Document]: List of Document objects, each containing a chunk of text
                            and associated metadata
            
        Note: Each Document has:
            - page_content: The actual text chunk
            - metadata: Dictionary with page number, total pages, and character count
        """
        # STEP 1: Load PDF using PyPDFLoader
        # PyPDFLoader reads the PDF and returns one Document per page
        # Each Document contains the page's text in page_content and page number in metadata
        loader = PyPDFLoader(pdf_path)
        pages = loader.load()
        
        # STEP 2: Process each page to create chunks
        processed_chunks = []  # Will store all chunks from all pages
        
        # Iterate through each page with its index
        # enumerate() gives us both the index (page_num) and the page Document
        for page_num, page in enumerate(pages):
            # 2a: Clean the raw text from the page
            # Removes extra spaces, newlines, and leading/trailing whitespace
            cleaned_text = self._clean_text(page.page_content)
            
            # 2b: Skip pages that are effectively empty
            # This prevents creating chunks from pages with minimal content
            # (e.g., pages with only page numbers or headers)
            if len(cleaned_text.strip()) < 50:
                continue
            
            # 2c: Split the cleaned text into chunks
            # create_documents() takes a list of texts and optional metadata
            # It applies the text_splitter's splitting logic to each text
            chunks = self.text_splitter.create_documents(
                texts=[cleaned_text],  # Single page text wrapped in a list
                metadatas=[{  # Metadata for this page (must match length of texts list)
                    "page": page_num + 1,  # Page number (1-indexed for readability)
                    "total_pages": len(pages),  # Total pages in the document
                    "char_count": len(cleaned_text)  # Character count of the raw page
                }]
            )
            
            # Add the chunks from this page to our master list
            processed_chunks.extend(chunks)
        
        # Return all chunks collected from all pages
        return processed_chunks
    
    def _clean_text(self, text: str) -> str:
        """
        Clean and normalize text extracted from PDF.
        
        PDFs often contain irregular spacing, multiple newlines, and other artifacts.
        This method normalizes the text to a consistent format.
        
        Steps:
        1. Split text by whitespace and rejoin with single spaces
           This removes all extra spaces, tabs, and newlines
        2. Strip leading and trailing whitespace
        
        Args:
            text (str): Raw text from PDF page
            
        Returns:
            str: Cleaned text with normalized whitespace
        
        Example:
            "Hello   world!\n\nHow are you?" → "Hello world! How are you?"
        """
        # " ".join(text.split()) does the following:
        # text.split() splits on any whitespace and returns a list of words
        # " ".join() joins them back with a single space between each word
        # This effectively collapses all multiple spaces, newlines, tabs, etc. to single spaces
        text = " ".join(text.split())
        
        # Remove leading and trailing whitespace
        text = text.strip()
        
        return text


# ============== USAGE EXAMPLE ==============
# This section demonstrates how to use the TextExtractor class

# Define the path to your PDF file
# Using raw string (r"") to handle Windows backslashes properly
pdf_path = r"C:\Users\Manish\OneDrive\Desktop\MANISH_2026\AI_WORKSETUP\langchain_init_proj\Actual_given_DataIngestParsing\0-DataIngestParsing\data\pdf\attention.pdf"

# Create an instance of TextExtractor with default parameters
# You can customize chunk_size and chunk_overlap if needed
text_extractor = TextExtractor()  # Uses chunk_size=1000, chunk_overlap=100

# Extract and chunk the text from the PDF
# This returns a list of Document objects, each containing a chunk of text
text_docs = text_extractor.extract_text(pdf_path)

# Print summary information about the extraction
print(f"Loaded {len(text_docs)} text chunks")

# If we have at least one chunk, display the first chunk's content and metadata
if text_docs:
    # Print first 100 characters of the first chunk's content
    # This gives a preview of what was extracted
    print(f"Page 1 content: {text_docs[0].page_content[:100]}...")
    
    # Print the metadata for the first chunk
    # This helps verify that page numbers and other info are correctly attached
    print(f"Metadata: {text_docs[0].metadata}")
```

```python
"""
EXTRACTOR 2: IMAGE EXTRACTION ONLY

This module is responsible for extracting images from PDF files.
It uses PyMuPDF (fitz) which provides robust image extraction capabilities
from PDF documents, preserving image formats and metadata.
"""

import fitz  # PyMuPDF - powerful PDF library for extracting images, text, and more
from PIL import Image  # Python Imaging Library for handling image data
from io import BytesIO  # For handling binary image data as file-like objects
from typing import List, Dict  # Type hints for better code documentation

class ImageExtractor:
    """
    A class to extract images from PDF files.
    
    This extractor scans through each page of a PDF, identifies embedded images,
    and extracts them along with their metadata (page number, format, size, etc.).
    
    The extraction process:
    1. Open the PDF using fitz (PyMuPDF)
    2. For each page, get all images embedded in that page
    3. For each image, extract the raw binary data and convert to PIL Image
    4. Store the image along with its metadata in a dictionary
    
    Attributes:
        None - this class is stateless and each extraction is independent
    """
    
    def extract_images(self, pdf_path: str) -> List[Dict]:
        """
        Extract all images from a PDF file and return them with metadata.
        
        The extraction process:
        1. Open the PDF document using fitz
        2. Iterate through each page
        3. For each page, get the list of images
        4. For each image:
           a. Get the XREF (cross-reference) number - this is how PDFs reference objects
           b. Extract the raw image data using the XREF
           c. Convert the raw bytes to a PIL Image object
           d. Store image data and metadata in a dictionary
        5. Return all image dictionaries in a list
        
        Args:
            pdf_path (str): File path to the PDF document
            
        Returns:
            List[Dict]: List of dictionaries, each containing:
                - page (int): Page number where image appears (1-indexed)
                - image_index (int): Index of image on that page (1-indexed)
                - format (str): Image file extension (e.g., 'png', 'jpeg')
                - size (int): Size of image data in bytes
                - pil_image (PIL.Image): The actual image object for further processing
        
        Note on PDF structure:
            - Each image in a PDF is stored as an object with a unique XREF number
            - fitz.extract_image(xref) returns a dictionary with:
                - "image": raw bytes of the image
                - "ext": file extension (format)
                - "width": image width in pixels
                - "height": image height in pixels
                - "colorspace": colorspace of the image
            - The PIL Image object allows for easy manipulation, display, or saving
        """
        # Initialize an empty list to store all extracted image data
        images_data = []
        
        # STEP 1: Open the PDF document using fitz (PyMuPDF)
        # Using 'with' statement ensures the document is properly closed after processing
        # This is important for memory management, especially with large PDFs
        with fitz.open(pdf_path) as doc:
            
            # STEP 2: Iterate through each page of the PDF
            # range(len(doc)) gives us page indices from 0 to total_pages-1
            for page_num in range(len(doc)):
                
                # Get the page object for the current page number
                # fitz pages are 0-indexed internally
                page = doc[page_num]
                
                # STEP 3: Get all images on this page
                # page.get_images() returns a list of image references
                # Each reference is a tuple where the first element is the XREF
                # XREF (cross-reference) is a unique identifier for the image object in the PDF
                image_list = page.get_images()
                
                # STEP 4: Process each image found on the page
                # enumerate gives us the image index (starting from 0) and the image tuple
                for img_idx, img in enumerate(image_list):
                    try:
                        # 4a: Extract the XREF from the image tuple
                        # The XREF is the first element and acts as a pointer to the image data
                        xref = img[0]
                        
                        # 4b: Extract the actual image data using the XREF
                        # doc.extract_image() returns a dictionary containing:
                        # - "image": bytes of the image
                        # - "ext": file extension
                        # - "width", "height": dimensions
                        # - "colorspace": color mode
                        base_image = doc.extract_image(xref)
                        
                        # 4c: Convert raw bytes to a PIL Image object
                        # BytesIO creates a file-like object from the bytes data
                        # Image.open() reads the image data and creates a PIL Image object
                        # This allows us to work with the image (resize, convert, etc.)
                        pil_image = Image.open(BytesIO(base_image["image"]))
                        
                        # 4d: Store the image and its metadata in a dictionary
                        # Using a dictionary provides a structured way to access both image and metadata
                        images_data.append({
                            "page": page_num + 1,  # Convert to 1-indexed for readability
                            "image_index": img_idx + 1,  # Convert to 1-indexed
                            "format": base_image["ext"],  # Image format (e.g., 'png', 'jpeg')
                            "size": len(base_image["image"]),  # Size in bytes
                            "pil_image": pil_image  # The actual PIL Image object
                        })
                        
                    except Exception as e:
                        # Error handling: log any issues but continue processing
                        # This ensures one bad image doesn't stop the entire extraction
                        print(f"Error on page {page_num+1}: {e}")
        
        # Return the complete list of extracted images with their metadata
        return images_data


# ============== USAGE EXAMPLE ==============
# This section demonstrates how to use the ImageExtractor class

# Define the path to your PDF file
# Using raw string (r"") to handle Windows backslashes properly
pdf_path = r"C:\Users\Manish\OneDrive\Desktop\MANISH_2026\AI_WORKSETUP\langchain_init_proj\Actual_given_DataIngestParsing\0-DataIngestParsing\data\pdf\attention.pdf"

# Create an instance of ImageExtractor
# No parameters needed as this class is stateless
image_extractor = ImageExtractor()

# Extract all images from the PDF
# This returns a list of dictionaries, each containing an image and its metadata
images = image_extractor.extract_images(pdf_path)

# Print summary information about the extraction
print(f"Loaded {len(images)} images")

# If we have at least one image, display its metadata
if images:
    # Access the first image's metadata
    # This shows us where the image was found and its basic properties
    print(f"Page {images[0]['page']}, Image {images[0]['image_index']}")
    print(f"Format: {images[0]['format']}")  # File format (PNG, JPEG, etc.)
    print(f"Size: {images[0]['size']} bytes")  # File size

    # Note: To save or display the image, you would use the 'pil_image' key:
    # images[0]['pil_image'].save('extracted_image.png')  # Save to file
    # images[0]['pil_image'].show()  # Display the image
```

# HANDLING PDF CHALLENGES

🔴 Purpose of This Section PDFs are notoriously difficult to parse because they:
  - Store text in complex ways (not just simple text)
  - Can have formatting issues

```python
# ============================================================================
# HANDLING PDF CHALLENGES
# ============================================================================
# PDFs are notoriously difficult to parse because they:
#   - Store text in complex ways (not just simple text)
#   - Can have formatting issues
#   - May contain scanned images (requiring OCR)
#   - Often have extraction artifacts
# ============================================================================

# ============================================================================
# EXAMPLE OF RAW PDF EXTRACTION
# ============================================================================
# This shows what raw text from a PDF looks like before cleaning
# Notice the excessive whitespace, newlines, and formatting artifacts
# ============================================================================

raw_pdf_text = """Company Financial Report

The financial performance for fiscal year 2023 shows significant growth in profitability.

Revenue increased by 25%.

The company's efficiency improved due to workflow optimization.

Page 1 of 10
"""

print("=" * 80)
print("RAW PDF EXTRACTION (BEFORE CLEANING):")
print("=" * 80)
print(repr(raw_pdf_text))
print("\n" + "-" * 80)

# ============================================================================
# PDF CLEANING FUNCTION
# ============================================================================
# This function addresses common PDF extraction issues:
# 1. Excessive whitespace (multiple spaces, newlines, tabs)
# 2. Ligature issues (common in PDF text extraction)
# 3. Formatting artifacts
# ============================================================================

def clean_text(text):
    """
    Clean raw PDF text by removing common extraction artifacts.
    
    Args:
        text (str): Raw text extracted from PDF
        
    Returns:
        str: Cleaned text with normalized whitespace and fixed ligatures
    """
    
    # Remove excessive whitespace
    # This converts multiple spaces, newlines, and tabs into single spaces
    # .split() splits on any whitespace and removes empty strings
    # " ".join() joins them back with single spaces
    text = " ".join(text.split())
    
    # Fix ligatures - common issue in PDF text extraction
    # Ligatures are characters that are combined into a single glyph
    # Common ligatures: fi, fl, ff, ffi, ffl, etc.
    # When extracted, they may appear as special characters or broken text
    text = text.replace("&", "fi")   # Fix 'fi' ligature
    text = text.replace("&", "fl")   # Fix 'fl' ligature
    # Note: The actual ligature characters may vary depending on the PDF
    
    return text

# ============================================================================
# APPLY THE CLEANING FUNCTION
# ============================================================================

cleaned = clean_text(raw_pdf_text)

print("=" * 80)
print("PDF TEXT CLEANING DEMONSTRATION:")
print("=" * 80)

print("\nBEFORE CLEANING:")
print("-" * 80)
print(repr(raw_pdf_text[:100]))  # Show first 100 characters
print("\nAFTER CLEANING:")
print("-" * 80)
print(repr(cleaned[:100]))       # Show first 100 characters

print("\n" + "=" * 80)
print("OUTPUT:")
print("=" * 80)

print("BEFORE:")
print(repr(raw_pdf_text[:100]))
print("\nAFTER:")
print(repr(cleaned[:100]))

# ============================================================================
# COMPARISON OF BEFORE AND AFTER
# ============================================================================

print("\n" + "=" * 80)
print("COMPARISON OF BEFORE AND AFTER CLEANING:")
print("=" * 80)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│  BEFORE CLEANING:                                                          │
│  ─────────────────────────────────────────────────────────────────────────  │
│  'Company Financial Report\\n\\n\\n The financial performance for fiscal   │
│  year 2024\\n shows significant gro'                                       │
│                                                                             │
│  Issues:                                                                    │
│  ❌ Multiple newlines (\\n\\n\\n)                                          │
│  ❌ Extra spaces (leading space)                                            │
│  ❌ Newline in middle of sentence (\\n shows)                              │
│  ❌ Inconsistent formatting                                                 │
│                                                                             │
│  AFTER CLEANING:                                                           │
│  ─────────────────────────────────────────────────────────────────────────  │
│  'Company Financial Report The financial performance for fiscal year       │
│  2024 shows significant growth in'                                         │
│                                                                             │
│  Improvements:                                                              │
│  ✅ Single spaces between all words                                        │
│  ✅ No leading/trailing whitespace                                         │
│  ✅ Text flows naturally without breaks                                    │
│  ✅ Consistent formatting                                                  │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ============================================================================
# COMMON PDF EXTRACTION ISSUES AND SOLUTIONS
# ============================================================================

print("=" * 80)
print("COMMON PDF EXTRACTION ISSUES AND SOLUTIONS:")
print("=" * 80)

print("""
┌─────────────────────────────────────────────────────────────────────────────┐
│  ISSUE 1: EXCESSIVE WHITESPACE                                             │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: PDFs often have multiple spaces, newlines, and tabs              │
│  Solution: text = " ".join(text.split())                                   │
│  Example: "Hello    World\\n\\n\\nHow are you?" → "Hello World How are you?"│
│                                                                             │
│  ISSUE 2: LIGATURE ARTIFACTS                                                │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Common ligatures (fi, fl, ff) are often corrupted in extraction │
│  Solution: text = text.replace("&", "fi")                                  │
│  Example: "ﬁnancial" → "financial"                                         │
│                                                                             │
│  ISSUE 3: PAGE NUMBERS AND HEADERS                                         │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Page numbers, headers, footers get mixed with content           │
│  Solution: Use regex to remove patterns like "Page X of Y"                │
│  import re                                                                 │
│  text = re.sub(r'Page \\d+ of \\d+', '', text)                            │
│                                                                             │
│  ISSUE 4: HYPHENATION                                                       │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Words may be hyphenated across lines                             │
│  Solution: text = text.replace("-\\n", "")  # Join hyphenated words        │
│  Example: "perfor-\\nmance" → "performance"                                │
│                                                                             │
│  ISSUE 5: SPECIAL CHARACTERS                                                │
│  ─────────────────────────────────────────────────────────────────────────  │
│  Problem: Em-dashes, en-dashes, curly quotes get converted to symbols     │
│  Solution: text = text.replace("–", "-")  # Convert to standard            │
│  text = text.replace("’", "'")   # Convert curly quotes                    │
│  text = text.replace("“", '"')   # Convert curly quotes                    │
│  text = text.replace("”", '"')   # Convert curly quotes                    │
└─────────────────────────────────────────────────────────────────────────────┘
""")

# ============================================================================
# ADVANCED CLEANING FUNCTION
# ============================================================================

print("=" * 80)
print("ADVANCED CLEANING FUNCTION:")
print("=" * 80)

print("""
Here's a more comprehensive cleaning function that handles multiple issues:

def advanced_clean_text(text):
    '''Comprehensive PDF text cleaning.'''
    import re
    
    # 1. Remove excessive whitespace
    text = " ".join(text.split())
    
    # 2. Fix common ligatures
    text = text.replace("&", "fi")
    text = text.replace("&", "fl")
    text = text.replace("&", "ff")
    
    # 3. Remove page numbers and headers
    text = re.sub(r'Page \\d+ of \\d+', '', text)
    text = re.sub(r'Page \\d+', '', text)
    
    # 4. Fix hyphenation (remove hyphens at line breaks)
    text = re.sub(r'(\\w+)-\\s+(\\w+)', r'\\1\\2', text)
    
    # 5. Fix common special characters
    text = text.replace("–", "-")
    text = text.replace("—", "-")
    text = text.replace("’", "'")
    text = text.replace("‘", "'")
    text = text.replace("“", '"')
    text = text.replace("”", '"')
    
    # 6. Remove multiple spaces again (just in case)
    text = " ".join(text.split())
    
    return text
""")

# ============================================================================
# SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("SUMMARY: PDF CLEANING BEST PRACTICES")
print("=" * 80)

print("""
✅ ALWAYS CLEAN PDF TEXTS BEFORE USE:
   → Raw PDF text is messy and inconsistent
   → Cleaning improves retrieval quality
   → Clean text is easier for LLMs to process

✅ START WITH SIMPLE CLEANING:
   → Remove excessive whitespace
   → Fix ligatures
   → Then handle specific issues

✅ KNOW YOUR PDF SOURCE:
   → Different PDF generators produce different artifacts
   → Test on a sample first
   → Adjust cleaning based on the output

✅ COMMON PDF CLEANING STEPS:
   1. Normalize whitespace
   2. Fix ligatures
   3. Remove page numbers
   4. Fix hyphenation
   5. Clean special characters
   6. Remove headers/footers

✅ FOR RAG APPLICATIONS:
   → Clean text before chunking
   → Clean text before embedding
   → Better input = Better output
""")

print("\n" + "=" * 80)
print("✅ PDF CLEANING DEMONSTRATION COMPLETE!")
print("=" * 80)
```

```python
from typing import List
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.schema import Document

class SmartPDFProcessor:
    """Advanced PDF processing with error handling"""
    def __init__(self, chunk_size=1000, chunk_overlap=100):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            separators=[" "],
        )

    def process_pdf(self, pdf_path: str) -> List[Document]:
        """Process PDF with smart chunking and metadata enhancement"""
        
        try:
            # Load PDF
            loader = PyPDFLoader(pdf_path)
            pages = loader.load()
            
            # Process each page
            processed_chunks = []
            
            for page_num, page in enumerate(pages):
                # Clean text
                cleaned_text = self._clean_text(page.page_content)
                
                # Skip empty pages
                if len(cleaned_text.strip()) == 0:
                    print(f"Skipping empty page {page_num+1}")
                    continue
                
                # Create chunks with enhanced metadata
                chunks = self.text_splitter.split_text(cleaned_text)
                
                # Create Document objects with metadata
                for chunk in chunks:
                    doc = Document(
                        page_content=chunk,
                        metadata={
                            **page.metadata,
                            "page": page_num + 1,
                            "total_pages": len(pages),
                            "chunk_method": "smart_pdf_processor",
                            "char_count": len(chunk)
                        }
                    )
                    processed_chunks.append(doc)
            
            return processed_chunks
            
        except Exception as e:
            print(f"Error loading PDF: {e}")
            return []

    def _clean_text(self, text: str) -> str:
        """Clean extracted text"""
        # Remove excessive whitespace
        text = " ".join(text.split())
        return text
```
