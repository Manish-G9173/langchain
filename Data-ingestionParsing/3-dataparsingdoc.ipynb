```python
# Import the Docx2txtLoader from langchain community
from langchain_community.document_loaders import Docx2txtLoader

print(" Using Docx2txtLoader")
print("=" * 50)
print("This method uses the 'docx2txt' library to extract text from Word documents.")
print("It's the simplest and most reliable method for loading .docx files.")
print("=" * 50)

try:
    # Create a loader instance by providing the file path
    # Docx2txtLoader reads the Word document and extracts all text content
    docx_loader = Docx2txtLoader("d:/word_files/proposal.docx")
    
    # The load() method reads the document and returns a list of Document objects
    # Each Document object contains:
    #   - page_content: The actual text content
    #   - metadata: Information about the document (source, file path, etc.)
    docs = docx_loader.load()
    
    # Display loading results
    print(f" Successfully loaded {len(docs)} document(s)")
    print(f" Content preview (first 200 characters): {docs[0].page_content[:200]}...")
    print(f" Document metadata: {docs[0].metadata}")
    print(f" Total characters in document: {len(docs[0].page_content)}")
    
    # You can now use the loaded content for further processing
    # For example, text splitting, embedding, or RAG applications
    
except FileNotFoundError:
    print(" File not found! Please check the file path.")
    print("   Suggestion: Ensure the file exists at 'd:/word_files/proposal.docx'")
except Exception as e:
    print(f" Error loading document: {e}")
    print("   Suggestion: Check if the file is corrupted or has proper permissions")
```

```python
##method 2 : unstructured 


# Import the UnstructuredWordDocumentLoader
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

print("\n Using UnstructuredWordDocumentLoader")
print("=" * 50)
print("This method uses the 'unstructured' library which provides more features:")
print("  - Preserves document structure (headers, paragraphs, lists)")
print("  - Can extract elements individually (mode='elements')")
print("  - More detailed metadata extraction")
print("=" * 50)

try:
    # Create loader with file path
    # mode='elements' tells the loader to return each document element separately
    # Elements can be: Title, NarrativeText, ListItem, Table, etc.
    loader = UnstructuredWordDocumentLoader(
        "data/word_files/proposal.docx", 
        mode="elements"  # 'elements' mode gives more granular content
        # mode="single" (default) returns the whole document as one chunk
    )
    
    # Load the document - this will return multiple Document objects
    # Each representing a different element in the document
    docs = loader.load()
    
    print(f"📦 Successfully loaded {len(docs)} document elements")
    
    # Display information about each element
    for i, doc in enumerate(docs[:5]):  # Show first 5 elements
        print(f"\n   Element {i+1}:")
        print(f"   - Type: {doc.metadata.get('category', 'Unknown')}")
        print(f"   - Content length: {len(doc.page_content)} characters")
        print(f"   - Preview: {doc.page_content[:100]}...")
    
    # Access specific metadata
    if docs:
        print(f"\n📄 First element metadata:")
        print(f"   - Source: {docs[0].metadata.get('source', 'N/A')}")
        print(f"   - Category: {docs[0].metadata.get('category', 'N/A')}")
        print(f"   - Element ID: {docs[0].metadata.get('element_id', 'N/A')}")
    
    # This is useful for:
    # - Processing structured documents
    # - Maintaining document hierarchy
    # - Identifying different content types (tables, lists, etc.)

except FileNotFoundError:
    print(" File not found! Please check the path.")
    print("   Suggestion: Verify 'data/word_files/proposal.docx' exists")
except Exception as e:
    print(f" Error loading document: {e}")
    print("   Suggestion: Try installing 'unstructured' library: pip install unstructured")
```
